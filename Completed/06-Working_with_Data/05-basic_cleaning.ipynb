# 🧼 Cleaning Dirty Data (Missing Values & Type Fixes)

## 🔹 LEARNING GOALS:
- Detect and count missing values (`NaN`)
- Fill or drop missing data
- Convert column data types safely
- Understand the difference between `NaN`, `None`, `""`, and type mismatches


### 🧪 1. Load a Messy Dataset

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Alice", "Bob", "Charlie", "David", None],
    "Age": ["25", "thirty", 35, np.nan, "40"],
    "Signup Date": ["2022-01-01", "not a date", "2022/03/01", None, "April 5, 2022"],
    "Score": [95.5, None, 88.0, 92.5, ""]
}

df = pd.DataFrame(data)
df
```

### 🧯 2. Detecting Missing or Broken Values

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
df[df.isnull().any(axis=1)]
```

### 🧹 3. Cleaning Strategy Options

```python
df.fillna({
    "Name": "Unknown",
    "Age": -1,
    "Signup Date": "1970-01-01",
    "Score": 0.0
})
```

```python
df.dropna()
```

### 🧬 4. Data Type Fixes

```python
df.dtypes
```

```python
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df["Score"] = pd.to_numeric(df["Score"], errors="coerce")
df["Signup Date"] = pd.to_datetime(df["Signup Date"], errors="coerce")
df.dtypes
```

### 🩹 5. Impute (Fill In) Fixed Missing Values

```python
df["Age"].fillna(df["Age"].median(), inplace=True)
df["Score"].fillna(df["Score"].mean(), inplace=True)
```

```python
df["Signup Date"].fillna(df["Signup Date"].min(), inplace=True)
```

```python
df["Name"].fillna("Unknown", inplace=True)
```

### 🤓 6. Cleaned Data Review

```python
df.info()
```

```python
df.describe(include="all")
```

### 🧪 Try It Yourself

Modify the `data` dictionary at the top of this notebook. Add:
- A new column with some `None` and `""` values
- At least one row with all columns filled incorrectly
Then re-run the notebook and fix it step-by-step.

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Alice", "Bob", "Charlie", "David", None, "Frank"],
    "Age": ["25", "thirty", 35, np.nan, "40", -5],
    "Signup Date": ["2022-01-01", "not a date", "2022/03/01", None, "April 5, 2022", ""],
    "Score": [95.5, None, 88.0, 92.5, "", 105], 
    "Status": ["Active", "Inactive", None, "", "Active", "Active"]
}

df_messy = pd.DataFrame(data)

df_messy.replace("", np.nan, inplace=True)

df_messy["Age"] = pd.to_numeric(df_messy["Age"], errors="coerce")
df_messy["Score"] = pd.to_numeric(df_messy["Score"], errors="coerce")
df_messy["Signup Date"] = pd.to_datetime(df_messy["Signup Date"], errors="coerce")


df_messy["Name"].fillna("Unknown", inplace=True)
df_messy["Status"].fillna("Unknown", inplace=True)
df_messy["Age"].fillna(df_messy["Age"].median(), inplace=True)

print("Cleaned Custom Data:\n", df_messy)
```

### 🧠 Mini-Challenge

> 🗂 Load `"data/survey.csv"` and:
- Identify which columns have missing values
- Use `.isnull().sum()` to get a null report
- Use a mix of `.fillna()`, `.dropna()`, and `pd.to_numeric()` or `pd.to_datetime()` to clean it
- Print a summary with `.info()` and `.describe()`

```python
survey_data = {
    "survey_id": [101, 102, 103, 104, 105],
    "rating": [5, np.nan, 4, 2, None],
    "comments": ["Great!", "", "Okay", None, "Bad"],
    "date": ["2023-01-01", "2023-01-02", "Invalid", "2023-01-04", "2023-01-05"]
}
survey = pd.DataFrame(survey_data)

print("\nNull Report:\n", survey.isnull().sum())

survey["date"] = pd.to_datetime(survey["date"], errors="coerce")
survey.dropna(subset=["date"], inplace=True)
survey["rating"].fillna(survey["rating"].median(), inplace=True)
survey["comments"].fillna("No Comment", inplace=True)

print("\nSurvey Info:")
survey.info()
```

### 📝 Summary
